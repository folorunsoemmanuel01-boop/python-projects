```python
import random

def hangman():

    words = ["python", "fish"]
    word = random.choice(words)

    play = 0
    print("Welcome to the HANGMAN GAME")

    while play < 3:
        guess = input("Enter a character")
        if guess.isalpha() and len(guess) == 1:
            print(guess)
            if guess in word:
                print(f"The position of the guess is {word.find(guess)}")
                print(f"Yes, The character {guess} is in {word}")
            else:
                print("No")

        else:
            print("Invalid chaaracter")
            
hangman()
        
```
