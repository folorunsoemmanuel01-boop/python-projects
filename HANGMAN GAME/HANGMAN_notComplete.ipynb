```python
import random

def randomizer():
    words = ["france", "argentina"]
    word = random.choice(words)

    play = 0
    print("Welcome to my HANGMAN game")

    while play < 4:
        guess = input("Enter a character: ")
        print(f"{guess}")

        if guess.isalpha() and len(guess) == 1:
            print(guess)
            if guess in word:
                print(f"The character is in position {word.find(guess)}")
                print(f"Yes, The character is in {word}")
            else:
                print("No")
        else:
            print("Invalid character")

randomizer()
```

```python
def callFunc(message):
    vari1 = int(input("Position of the character u want to change: ")) # this takes input of the position of d character u want to change
    vari2 = input("Desired character: ") # takes the input of our desired character
    liste = list(message)
    liste[vari1] = vari2
    return liste # gives back the latest value of variable "liste"

callFunc("Median") # call back the function declared
```
