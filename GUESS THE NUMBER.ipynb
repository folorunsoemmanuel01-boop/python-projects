```python
import random

def hangman():

    # List of words
    words = ["python", "apple", "school", "computer", "football", "student"]

    # Select  random word
    word = random.choice(words)

    #  list of underscores
    progress = ["_"] * len(word)


    print(" WELCOME TO HANGMAN ")

    while True:

        # Display  progress
        print("\nWord:", " ".join(progress))

        # Get user input
        guess = input("Guess a letter: ").lower()

        # Check if letter exists
        if guess in word:

            # Update all matching positions
            for i in range(len(word)):
                if word[i] == guess:
                    progress[i] = guess

            print("Correct!")

        else:
            print("Wrong guess!")

        # Check if player has won
        if "_" not in progress:
            print("\nCongratulations!")
            print("The word was:", word)
            print("You Win!")
            break
    

# Call the function
hangman()
```
