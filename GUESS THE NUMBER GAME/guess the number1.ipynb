```python
import random
import time

def guessing_number():
    randomNumber = random.randint(1, 20)
    print("WELCOME to GUESSING THE NUMBER game.")
    name = input("Enter your name: ")
    print(f"Hola!, {name}")
    print("Have fun")
    

    attempts = 10
    play = True

    score_sheet = {}

    while play:
        guessing_number = int(input("Enter a number"))
        if guessing_number > randomNumber:
            attempts -= 1
           
            time.sleep(1)
            print("Too High")

        elif guessing_number < randomNumber:
            attempts -= 1
            
            time.sleep(1)
            print("Too Low")

        else:
            time.sleep(1)
            print(f"Congratulations {name}, You got the number!")
            

            print("\n")
            decision = input("Do you want to continue? Yes or No")
            if decision.lower() == 'yes':
                attempts = 10
                randomNumber = random.randint(1, 20)

            else:
                play = False
                time.sleep(1)
                print(f"Thank u 4 playing {name} ")
                print(score_sheet)
                
        
            print(f"You have {attempts} attempts left!")
            if attempts == 0:
                score_sheet[name] = "You failed!"
                

guessing_number()
```
