```python
import random
import time

def guess_number():
    randomNumber = random.randint(1, 20)
    print("Welcome to GUESS THE NUMBER game")
    name = input("Enter your name: ")
    print(f"Welcome, {name}")

    attempts = 10
    play = True

    score_sheet = {}

    while play:
        guess_number = int(input("Enter a number: "))
        if guess_number > randomNumber:
            attempts -= 1
            time.sleep(1)
            print("Too High")
            print(f"You have {attempts} attempts left")

        elif guess_number < randomNumber:
            attempts -= 1
            time.sleep(1)
            print("Too Low")
            print(f"You have {attempts} attempts left")

        else:
            print(f"You got the number {name}")
            score_sheet[name] = 'completed'
            

            print("\n")
            decision = input("Do you want to continue? Yes or No ")
            if decision.lower() == 'yes':
                attempts = 10
                time.sleep(1)
                randomNumber = random.randint(1, 20)
                name = input("Enter your name: ")
                print(f"Welcome {name}")

            else:
                play = False
                time.sleep(1)
                print(score_sheet)
                print("Thank u 4 playing")
                

            if attempts == 0:
                score_sheet[name] = 'failed'
                print(score_sheet)
                break

guess_number()
```
